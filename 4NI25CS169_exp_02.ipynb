```python
import pandas as pd
import numpy as np
```

```python
data ={
    'name': ['Amit','bebo','Chetan','Amit','Divya','Ravi'],
    'Age':[25,np.nan,30,25,28,22],
    'Marks': [85,90,np.nan,85,78,95],
    'city':['Mysore','Mysore','Bengaluru','Mysore',np.nan,'Delhi']
}
df=pd.DataFrame(data)
print("Dataset:")
display(df)
```

```python
print("\n Shape of datset (rows,columns):",df.shape)
print("\n Column names:",df.columns.tolist())
print("\n Data types of each colomn :")
print(df.dtypes)
print("\n Dataset imnfo:")
df.info()
```

```python
print("\n Descriptive statics (numerical columns):")
print(df.describe)
print("\n Mean :\n ",df.mean(numeric_only=True))
print("\n Meadian:\n ",df.median(numeric_only=True))
print("\n Standard Deviation:\n",df.std(numeric_only=True))
```

```python
## step 4: Identifying Missing Values
print("\n Missing values in each column:")
print(df.isnull().sum())

print("\n Total missing values in dataset :",df.isnull().sum().sum())
```

```python
# Step 5 :Identifying Duplicate Records
print("\nDuplicate rows in dataset:")
print(df[df.duplicated()])

print("\n Number of duplicate rows:",df.duplicated().sum())
print("\n EDA process completed succesfully!")
```
