```python
plt.figure(figsize=(8,5))
plt.hist(df["Final_Score"],bins=10,edgecolor="black")
plt.title("Distribution of the Final Score")
plt.xlabel("Final Score");plt.ylabel("Number of Students")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(y=df["Final_Score"])
plt.title("Box plot of the Final Score")

plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Study_Hours"],df["Final_Score"])
plt.title("Study hours vs Final Score")
plt.xlabel("Study Hours");plt.ylabel("Final Score")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df,x="Study_Hours",y="Final_Score",hue="Department",s=80)
plt.title("Study hours vs Final Score")
plt.xlabel("Study Hours");plt.ylabel("Final Score")
plt.show()
```

```python
num_cols=["Attendance","Study_Hours","Assignment_Score","Internal_Marks","Final_Score"]
sns.pairplot(df[num_cols])
plt.show()
```

```python
plt.figure(figsize=(9,6))
sns.heatmap(df[num_cols].corr(),annot=True,fmt=".2f",cmap="coolwarm")
plt.title("Correlation Heatmap")

plt.show()
```

```python
movie_data={"Movie":["Sky Force","Code Red","Midnight Run","Ocean Echo","City Lights","Last Game","Silent Road","Future World","Hidden Truth","Wild Journey","Digital Dreams","Final Mission"],"Budget_Crore":[80,120,60,45,35,70,50,150,40,90,110,130],"Duration_Minutes":[145,160,125,135,110,150,130,170,120,155,165,158],"Rating":[8.2,7.5,8,7.8,6.9,8.5,7.6,7.2,8.1,7.9,7.4,8.3],"Revenue_Crore":[240,280,150,110,75,260,130,310,140,220,290,350]}
movies=pd.DataFrame(movie_data)
movies
```

## Questions for above  data

1. Histogram of rating
2. Box plot of Budget_crore
3. Scatter plot:Budget_crore vs Revenue_crore
4. Pair plot for numerical columns
5. Correlation heatmap
6. Write three data-driven observations.


```python
plt.figure(figsize=(8,5))
plt.hist(movies['Rating'], bins=8, edgecolor='black', color='skyblue')
plt.title('Histogram of Movie Ratings')
plt.xlabel('Rating')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(y=movies['Budget_Crore'], color='lightgreen')
plt.title('Box Plot of Budget (Crore)')
plt.xlabel('Budget_Crore')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(movies['Budget_Crore'], movies['Revenue_Crore'], color='coral')
plt.title('Budget vs Revenue')
plt.xlabel('Budget (Crore)')
plt.ylabel('Revenue (Crore)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
sns.pairplot(movies[['Budget_Crore', 'Duration_Minutes', 'Rating', 'Revenue_Crore']])
plt.suptitle('Pair Plot of Numerical Movie Data', y=1.02)
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(movies[['Budget_Crore', 'Duration_Minutes', 'Rating', 'Revenue_Crore']].corr(), annot=True, cmap='YlGnBu')
plt.title('Correlation Heatmap')
plt.show()
```
